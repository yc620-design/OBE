```python
from sklearn.model_selection import train_test_split

# Step 2 - Dropping Unwanted Columns and Splitting the Dataset into train and test set
# Note: The Obesity dataset consists of highly relevant features, so dropping columns is not strictly necessary here.
# We split the cleaned dataset into 70% training data and 30% testing data.

# Define obe_cleaned from obe, as no explicit cleaning steps were performed to create a separate DataFrame.
obe_cleaned = obe

dataset_train, dataset_test = train_test_split(obe_cleaned, test_size=0.3, random_state=42)

# Print the results to show in your report
print(f"Total dataset size: {obe_cleaned.shape[0]} rows")
print(f"Training set size: {dataset_train.shape[0]} rows (70%)")
print(f"Testing set size: {dataset_test.shape[0]} rows (30%)")
```

```python
#Step 1: Visualizing Outliers using Boxplots
import matplotlib.pyplot as plt
import seaborn as sns

# Visualizing outliers using boxplots for continuous numerical features
plt.figure(figsize=(10, 6))

# We select only Age, Height, and Weight for the boxplot
sns.boxplot(data=obe_cleaned[['Age', 'Height', 'Weight']], palette="Set2")

plt.title('Figure: Boxplot for Detecting Outliers in Numerical Features')
plt.ylabel('Values')
plt.show()
```

```python
print("Duplicate rows before:", obe.duplicated().sum())

obe = obe.drop_duplicates()

print("Duplicate rows after:", obe.duplicated().sum())
```

```python
# Calculating Outliers for the 'Weight' column using the IQR Method
Q1 = obe_cleaned['Weight'].quantile(0.25)
Q3 = obe_cleaned['Weight'].quantile(0.75)
IQR = Q3 - Q1

# Define the lower and upper bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter the dataset to find the outlier rows
outliers_weight = obe_cleaned[(obe_cleaned['Weight'] < lower_bound) | (obe_cleaned['Weight'] > upper_bound)]

print("--- Outlier Detection using IQR ---")
print(f"Lower Bound for Weight: {lower_bound:.2f}")
print(f"Upper Bound for Weight: {upper_bound:.2f}")
print(f"Total number of statistical outliers in Weight: {outliers_weight.shape[0]} rows")

# Note for report:
# Since this is an Obesity dataset, "extreme" weights are actual target classes (e.g., Obesity Type III).
# Therefore, dropping these outliers would result in data loss for critical categories. We will keep them.
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Encode the target variable: Convert 'NObeyesdad' (Obesity Level) to numeric labels (0, 1, 2, 3...)
le = LabelEncoder()
obe_cleaned['Target'] = le.fit_transform(obe_cleaned['NObeyesdad'])

# Print the mapping dictionary for your report reference
obesity_mapping = dict(zip(le.classes_, le.transform(le.classes_)))
print("Numeric mapping for Obesity Levels:\n", obesity_mapping)

# 2. Separate features (X) and the target variable (y)
X = obe_cleaned.drop(columns=['NObeyesdad', 'Target'])
y = obe_cleaned['Target']

# 3. Apply One-Hot Encoding to categorical features (e.g., Gender, SMOKE)
# drop_first=True prevents the dummy variable trap (multicollinearity)
X_encoded = pd.get_dummies(X, drop_first=True)

print("\nFirst two rows of the encoded features (X_encoded):")
display(X_encoded.head(2))
```

```python
from sklearn.preprocessing import StandardScaler

# Initialize the StandardScaler
scaler = StandardScaler()

# Apply standardization to scale all features to a similar range
# (Mean = 0, Standard Deviation = 1)
X_scaled = pd.DataFrame(scaler.fit_transform(X_encoded), columns=X_encoded.columns)

print("Standardization complete! First two rows of scaled features (X_scaled):")
display(X_scaled.head(2))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

# Parameter tuning
param_grid_lr = {
    'C': [0.01, 0.1, 1, 10]
}

grid_lr = GridSearchCV(
    LogisticRegression(max_iter=1000, random_state=42),
    param_grid=param_grid_lr,
    cv=5,
    scoring='accuracy'
)

# Train model
grid_lr.fit(X_train_scaled, y_train)

# Best model
best_lr = grid_lr.best_estimator_

# Prediction
y_pred_lr = best_lr.predict(X_test_scaled)

print("Best Parameters:", grid_lr.best_params_)
print("Best Cross Validation Accuracy:", grid_lr.best_score_)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier

param_grid_knn = {
    'n_neighbors': [3, 5, 7, 9],
    'weights': ['uniform', 'distance']
}

grid_knn = GridSearchCV(
    KNeighborsClassifier(),
    param_grid=param_grid_knn,
    cv=5,
    scoring='accuracy'
)

grid_knn.fit(X_train_scaled, y_train)

best_knn = grid_knn.best_estimator_

y_pred_knn = best_knn.predict(X_test_scaled)

print("Best Parameters:", grid_knn.best_params_)
print("Best Cross Validation Accuracy:", grid_knn.best_score_)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

param_grid_svm = {
    'C': [0.1, 1, 10],
    'kernel': ['linear', 'rbf']
}

grid_svm = GridSearchCV(
    SVC(random_state=42),
    param_grid=param_grid_svm,
    cv=5,
    scoring='accuracy'
)

grid_svm.fit(X_train_scaled, y_train)

best_svm = grid_svm.best_estimator_

y_pred_svm = best_svm.predict(X_test_scaled)

print("Best Parameters:", grid_svm.best_params_)
print("Best Cross Validation Accuracy:", grid_svm.best_score_)
```

```python
from sklearn.metrics import accuracy_score

lr_acc = accuracy_score(y_test, y_pred_lr)
knn_acc = accuracy_score(y_test, y_pred_knn)
svm_acc = accuracy_score(y_test, y_pred_svm)

print("Logistic Regression:", lr_acc)
print("KNN:", knn_acc)
print("SVM:", svm_acc)
```

```python
import matplotlib.pyplot as plt

models = ['Logistic Regression', 'KNN', 'SVM']
accuracy = [lr_acc, knn_acc, svm_acc]

plt.figure(figsize=(8,5))

bars = plt.bar(models, accuracy)

plt.title("Accuracy Comparison")
plt.xlabel("Machine Learning Models")
plt.ylabel("Accuracy")
plt.ylim(0,1)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x()+bar.get_width()/2,
        height+0.01,
        f"{height:.3f}",
        ha='center'
    )

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, confusion_matrix

models = {
    'Logistic Regression (Baseline)': y_pred_lr,
    'K-Nearest Neighbors (KNN)': y_pred_knn,
    'Support Vector Machine (SVM)': y_pred_svm
}

# 1. Performance Summary Table
results = []
for name, y_pred in models.items():
    acc = accuracy_score(y_test, y_pred)
    prec, rec, f1, _ = precision_recall_fscore_support(y_test, y_pred, average='weighted')
    results.append({'Model': name, 'Accuracy': acc, 'Precision': prec, 'Recall': rec, 'F1-Score': f1})

df_results = pd.DataFrame(results)
print("=== Final Model Comparison ===")
display(df_results)

# 2. Confusion Matrix Heatmaps
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
class_names = le.classes_

for ax, (name, y_pred) in zip(axes, models.items()):
    cm = confusion_matrix(y_test, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=class_names, yticklabels=class_names)
    ax.set_title(f'{name}', fontsize=12, fontweight='bold')
    ax.set_xlabel('Predicted Label')
    ax.set_ylabel('True Label')
    ax.tick_params(axis='x', rotation=90)

plt.tight_layout()
plt.show()
```
